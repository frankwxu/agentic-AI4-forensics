# First AgentScope model call

This notebook sends one small request to the course Ollama endpoint. Confirm that `.env` contains `MODEL` and `OLLAMA_BASE_URL` before running it.

```python
import asyncio
import os

from dotenv import load_dotenv

from agentscope.credential import OpenAICredential
from agentscope.message import Msg, TextBlock
from agentscope.model import OpenAIChatModel
```

```python
load_dotenv()

model_name = os.getenv("MODEL")
base_url = os.getenv("OLLAMA_BASE_URL")
if not model_name or not base_url:
    raise RuntimeError("Set MODEL and OLLAMA_BASE_URL in .env before running this notebook.")

model = OpenAIChatModel(
    credential=OpenAICredential(api_key="ollama", base_url=base_url),
    model=model_name,
    stream=False,
    parameters=OpenAIChatModel.Parameters(temperature=0, max_tokens=30),
)
```

```python
response = await model(
    messages=[
        Msg(
            name="user",
            role="user",
            content=[TextBlock(text="Reply with exactly: AgentScope is ready.")],
        ),
    ],
)
print("".join(block.text for block in response.content if isinstance(block, TextBlock)))
```
