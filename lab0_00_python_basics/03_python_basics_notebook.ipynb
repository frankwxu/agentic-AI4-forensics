## 1. Variables and Simple Values

Variables give a value a name so later lines can reuse it. Start by reading the variable name, then the value after `=`.

```python
# A string stores text.
student_name = "Maya"
# An integer stores a whole number.
course_count = 3
# A float stores a number with a decimal point.
average_score = 91.5

print("Student:", student_name)
print("Courses:", course_count)
print("Average score:", average_score)
```

## 2. Lists and Dictionaries

Lists keep items in order. Dictionaries connect a key to a value. These are the two most common data shapes in the later notebooks.

```python
# Square brackets create a list: an ordered collection of items.
courses = ["Python Basics", "Data Skills", "AI Basics"]
# Curly braces create a dictionary: each key labels one stored value.
student = {"name": "Maya", "major": "Computer Science", "year": 2}

# List positions start at 0, so [0] means the first item.
print("First course:", courses[0])
# A dictionary key inside brackets retrieves its matching value.
print("Major:", student["major"])
print("Dictionary keys:", list(student))
```

## 5. Strings, f-Strings, and Boolean Logic

Later notebooks clean text, build readable status messages, and make decisions with `and`, `or`, and `not`. An f-string starts with `f` and places a variable inside `{braces}`.

```python
# .strip() removes extra spaces at the beginning and end.
raw_name = "  Maya  "
# Save the cleaned text in a new variable; raw_name stays unchanged.
clean_name = raw_name.strip()

# .lower() makes comparison text consistent.
day = "Monday".lower()
```

Now use the cleaned values in a Boolean decision and display the result with an f-string.

```python
# Each comparison produces True or False.
has_name = clean_name != ""
# `in` checks whether day appears in this set of allowed values.
is_weekday = day in {"monday", "tuesday", "wednesday", "thursday", "friday"}
# Combine the True/False values into one final decision.
ready_to_continue = has_name and is_weekday

# The f before the string inserts values from the braces.
print(f"Name: {clean_name}; day: {day}; ready: {ready_to_continue}")
```

**What to notice:** `and` means every condition must be true; `or` means at least one condition may be true; `not` reverses a true/false result. Read an f-string from left to right and replace each `{variable}` with its current value.

## 6. Safe Dictionary Lookup With `.get(...)` and `None`

Settings and configuration files may not contain every optional field. `.get(...)` safely returns `None` or a default value instead of stopping the notebook with a missing-key error.

```python
settings = {"student": "Maya"}
# This key may not be present in every settings dictionary.
difficulty = settings.get("difficulty")

# The second argument is a default when the key is missing.
student_label = settings.get("student", "not recorded")

# `is None` asks whether Python received no value.
# Follow one branch when no difficulty exists, otherwise use the other branch.
if difficulty is None:
    print(f"No difficulty was recorded for {student_label}.")
else:
    print(f"Difficulty for {student_label}: {difficulty}")
```

**What to notice:** Use `record["key"]` only when the key must exist. Use `record.get("key")` when missing information is possible and should be handled carefully.

## 8. Small Helper Functions

Later notebooks use small helpers to read a file once and return data in a consistent shape. This example is longer because it includes file reading; focus first on the function name, input, and return value.

```python
import csv
from pathlib import Path

data_dir = Path("data/python_basics")
# This helper accepts a Path and returns a list of row dictionaries.
def read_courses(path: Path) -> list[dict]:
    with path.open(encoding="utf-8", newline="") as handle:
        # The return value leaves the function immediately.
        return list(csv.DictReader(handle))

# A second helper returns a short, predictable list.
def make_study_plan() -> list[str]:
    return ["Python Basics", "Data Skills", "AI Basics"]

# Call both helpers and save their returned values in new variables.
rows = read_courses(data_dir / "courses.csv")
study_plan = make_study_plan()
study_plan
```

## 10. Dictionary Comprehensions, `enumerate(...)`, and Type Hints

A dictionary comprehension builds a lookup table. `enumerate(...)` gives a loop both a counter and an item. These patterns make later lookups and lists easier to read.

```python
course_rows = [{"course": "Data Skills"}, {"course": "Python Basics"}]
course_rows
```

Next, build a dictionary that uses each course name as a lookup key.

```python
# This dictionary comprehension uses each course name as a key for its whole row.
rows_by_course = {row["course"]: row for row in course_rows}
rows_by_course
```

Finally, use one course name to retrieve its row from the lookup table.

```python
course_name = "Data Skills"
rows_by_course[course_name]
```

```python
course_rows = [{"course": "Data Skills"}, {"course": "Python Basics"}]
# Type hints tell the reader that rows is a list of dictionaries.
def summarize_events(rows: list[dict]) -> dict:
    return {"count": len(rows), "first_course": rows[0]["course"]}

# Call the helper and print the dictionary it returns.
summarize_events(course_rows)
```

```python
# enumerate supplies both a human-friendly counter and the current list item.
for number, course_name in enumerate(sorted_courses, start=1):
    print(number, course_name)
```

## 11. Growing and Selecting Collections

Later notebooks build lists and select matching values. These small examples show one pattern at a time.

```python
course_rows = [{"course": "Data Skills"}, {"course": "Python Basics"}]
completed_courses = []  # Start with an empty list.

# [:2] means the first two rows. .append(...) adds one new item to the end.
for row in course_rows[:2]:
    # Add the current course to the end of the list.
    completed_courses.append(row["course"])

completed_courses
```

```python
# range(1, 4) produces 1, 2, 3.
for step in range(1, 4):
    print(f"Step {step}")
```

```python
course_rows = [{"course": "Data Skills"}, {"course": "Python Basics"}]
# next(...) returns the first matching row. The final None is a safe fallback.
python_row = next((row for row in course_rows if row["course"] == "Python Basics"), None)
python_row
```

**What to notice:** `(row for row in ...)` is a generator expression: it checks rows one at a time. `next(...)` stops at the first match, so always provide a fallback such as `None` when a match may not exist.

## 12. Error Handling With `try`, `except`, and `raise`

Later notebooks stop early when a folder, setting, or expected value is missing. Error handling makes the problem explicit instead of allowing confusing results later.

```python
settings = {}
# This helper returns a required value or stops with a named error.
def require_setting(settings: dict, field_name: str) -> str:
    # .get returns None if the field is absent.
    value = settings.get(field_name)
    if value is None:
        # raise stops this function and explains exactly what is missing.
        raise KeyError(f"Settings are missing required field: {field_name}")
    return value

try:
    # This practice field is intentionally absent, so the except block will run.
    require_setting(settings, "missing_practice_field")
except KeyError as error:
    # `error` stores the message created by raise so it can be shown clearly.
    print(f"Handled setup problem: {error}")
```

**What to notice:** `try` contains work that may fail. `except` handles one expected kind of failure. `raise` creates a clear error when code cannot safely continue.

## 13. Calling a Function With `**arguments`

A function call can receive its inputs from a dictionary. `**arguments` turns matching dictionary keys into named function inputs.

```python
def add(a, b):
    return a + b

# Use keyword arguments to name each input.
# Here, a=2 and b=3 are keyword arguments.
add(a=2, b=3)
```

The result is the same as add(2, 3); the names make each input explicit.

```python
arguments = {"a": 2, "b": 3}
arguments
```

Next, **arguments passes each dictionary value to add using its matching key.

```python
# **arguments is equivalent to add(a=2, b=3).
add(**arguments)
```

**What to notice:** The dictionary keys must match the function parameter names.

## 14. A Small `@tool` Example

Later course notebooks use `@tool`, so this is the one course-specific pattern included here. Read it as: wrap the following function so another program can discover its name, inputs, and output. You do not need decorator theory for this course.

```python
# Later notebooks import the full @tool helper from src/. This small version is only for syntax practice.
def tool(function):
    # Attach a readable name to the function object.
    function.name = function.__name__
    # Attach .run as another way to call the same function.
    function.run = function
    # Return the changed function so @tool can replace the original name.
    return function

# @tool runs the tool(...) function on say_hello after it is defined.
@tool
def say_hello(person: str) -> str:
    """Return a friendly greeting for one person."""
    # Return one f-string using the supplied name.
    return f"Hello, {person}!"

# Read the attribute added by the decorator.
print("Tool name:", say_hello.name)
# Call the function through its attached .run name with a keyword argument.
print("Tool result:", say_hello.run(person="Maya"))
```

## 15. Classes and Methods: A Reading Pattern

Later notebooks use classes to keep related data and behavior together. You do not need to design classes for this course. You only need to recognize that `self` means the particular object being used, and that a method such as `.show()` uses that object’s stored information.

```python
# A class is a blueprint for creating related objects.
class GreetingCard:
    # __init__ runs when Python creates a new GreetingCard object.
    def __init__(self, recipient: str, message: str):
        # Store the supplied values on this particular object.
        self.recipient = recipient
        self.message = message

    # A method is a function stored inside a class.
    def show(self) -> str:
        # self.recipient and self.message come from the object created below.
        return f"To {self.recipient}: {self.message}"

# Create one object, then read its attributes and call its method.
card = GreetingCard("Maya", "Have a great day!")
print("Recipient:", card.recipient)
print("Card message:", card.show())
```

**What to notice:** `card` is an object. It keeps `recipient` and `message` as attributes, and `.show()` is a method. Later classes follow the same read-the-object-then-call-a-method pattern.
